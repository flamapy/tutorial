## **Operations**

### Atomic Sets


Basic usage

```python
from flamapy.core.discover import DiscoverMetamodels
dm = DiscoverMetamodels()
result=dm.use_operation_from_file("AtomicSets","./samplemodel.uvl")
print(result)
```

Advanced usage

```python
from flamapy.core.discover import DiscoverMetamodels
dm = DiscoverMetamodels()
feature_model = dm.use_transformation_t2m("./samplemodel.uvl", 'fm')
result = dm.use_operation(feature_model,'FMAtomicSets').get_result()
print(result)
```

### Conflict detection

Basic usage

```python
from flamapy.core.discover import DiscoverMetamodels
# Initialize the discover metamodel
dm = DiscoverMetamodels()
# Call the operation. Transformations will be automatically executed
result = dm.use_operation_from_file("PySATConflict", "./samplemodel.uvl", configuration_file="sampleconfiguration.csvconf")
print(result)
```

Advanced usage

```python
from flamapy.core.discover import DiscoverMetamodels
# Initialize the discover metamodel
dm = DiscoverMetamodels()
# Get the fm metamodel representation using the transformation required to get to the fm metamodel
feature_model = dm.use_transformation_t2m("./samplemodel.uvl", 'fm')
# Get the configuration
configuraion = dm.use_transformation_t2m("sampleconfiguration.csvconf", 'configuration')
# Manually call a M2M transformation to Pysat
sat_model = dm.use_transformation_m2m(feature_model, "pysat_diagnosis")
# Get the operation
operation = dm.get_operation(sat_model, 'PySATConflict')
# Assign the configuration
operation.set_configuration(configuraion)
# Execute the operation
operation.execute(sat_model)
# Get and print the result
result = operation.get_result()
print(result)
```
