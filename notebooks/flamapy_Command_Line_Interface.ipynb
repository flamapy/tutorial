## Command Line Interface

**flamapy** as a tool aims at being easily usable; therefore, it provides a simple command line interface that does not require learning the framework and enables the most common feature modelling operations.

## Setup

Install the distribution

```python
!pip install flamapy
```

Download sample UVL and configuration

```python
!wget -q "https://raw.githubusercontent.com/flamapy/flamapy/main/resources/models/simple/valid_model.uvl" -O "samplemodel.uvl"
!wget -q "https://raw.githubusercontent.com/flamapy/flamapy/main/resources/configurations/valid_configuration.csvconf" -O "sampleconfiguration.csvconf"
```

Show sample model

```python
!cat "./samplemodel.uvl"
```

Show sample configuration

```python
!cat "sampleconfiguration.csvconf"
```

## Operations

Atomic Sets

```python
!flamapy atomic_sets "./samplemodel.uvl"
```

Average Branching Factor

```python
!flamapy average_branching_factor "./samplemodel.uvl"
```

Commonality

```python
!flamapy commonality "./samplemodel.uvl" "./sampleconfiguration.csvconf"
```

Configurations

```python
!flamapy configurations "./samplemodel.uvl"
```

Number of Configurations

```python
!flamapy configurations_number "./samplemodel.uvl"
```

Core Features

```python
!flamapy core_features "./samplemodel.uvl"
```

Count leafs

```python
!flamapy count_leafs "./samplemodel.uvl"
```

Dead Features

```python
!flamapy dead_features "./samplemodel.uvl"
```

Estimated number of configurations

```python
!flamapy estimated_number_of_configurations "./samplemodel.uvl"
```

False optional features

```python
!flamapy false_optional_features "./samplemodel.uvl"
```

Feature ancestors

```python
!flamapy feature_ancestors "./samplemodel.uvl" "PHP"
```

Filter

```python
!flamapy filter "./samplemodel.uvl" "./sampleconfiguration.csvconf"
```
